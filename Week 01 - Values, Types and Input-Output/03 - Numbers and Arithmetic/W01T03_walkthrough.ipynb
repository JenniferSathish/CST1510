## 2. `/` always gives a float

Even when it divides exactly.

```python
print(10 / 2)
print(type(10 / 2))
```

`//` throws away the fraction. `%` gives what is left over.

```python
print(10 // 3)
print(10 % 3)
```

Together they split a number into parts. This is more useful than it looks.

```python
total_minutes = 137

hours = total_minutes // 60
minutes = total_minutes % 60

print(total_minutes, "minutes is", hours, "hours and", minutes, "minutes")
```

One split is useful. Chaining two splits, from the biggest unit down, is what you actually need for a stopwatch or a duration.

```python
total_seconds = 4831

hours = total_seconds // 3600
remaining = total_seconds % 3600

minutes = remaining // 60
seconds = remaining % 60

print(total_seconds, "seconds =", hours, "hours,", minutes, "minutes,", seconds, "seconds")
```

Same idea twice: split off the biggest unit first with `// 3600` and `% 3600`, then split what is *left over* into minutes and seconds the same way you already did above.

## 3. Order of operations

Python follows normal maths rules: `**` first, then `*` `/` `//` `%`, then `+` `-`.

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
```

> **Habit worth forming:** use brackets even when you do not need them. They cost nothing and remove all doubt.

## 4. The calculation you will write most often

*What share of the whole is this?*

```python
value = 23.7
limit = 20

difference = value - limit
percent = (value / limit) * 100

print("Difference:", difference)
print("Percent of limit:", percent)
```

## 5. Shorthand

Changing a variable using its own value is so common it has a short form.

```python
count = 0

count = count + 1     # the long way
print(count)

count += 1            # exactly the same thing
print(count)

count += 5
print(count)
```

---
## Common mistakes

| Mistake | What happens | Fix |
|---|---|---|
| `2 + 3 * 4` when you meant `(2 + 3) * 4` | wrong answer, **no error** | use brackets |
| dividing by zero | `ZeroDivisionError` — program stops | check the total is not 0 |
| expecting `10 / 2` to be `5` | it is `5.0` | `/` always gives a float |
| `3.6999999999999993` | not a bug | format it: `f"{x:.2f}"` |
