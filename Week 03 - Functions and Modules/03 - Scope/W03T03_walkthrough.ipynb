## 1. Local means local

A variable created inside a function only exists inside it.

````python
def check():
    result = "OK"
    print(result)

check()

# print(result)   # NameError: name 'result' is not defined
````

```
outside
   │
   ├── calls check()
   │        │
   │        ▼
   │   result = "OK"    (created here, local only)
   │        │
   │        ▼
   │   check() ends — result stops existing
   │
   └── print(result) ──▶  NameError. It never existed out here.
```

````python
def double(n):
    n = n * 2
    return n

number = 21
print(double(number))   # 42
print(number)             # still 21 - the n inside double is its own copy
````

## 3. Reading an outer variable is fine. Changing it is not.

A function can **read** a variable defined outside it. It cannot change your copy of
it — only give you back a new value, which you choose whether to keep.

````python
def add_tax(price):
    price = price * 1.2   # only changes the LOCAL copy
    return price            # this is what actually gets it out

total = 100
total = add_tax(total)   # catch it and reassign
print(total)               # 120.0
````

## 4. Returning more than one value

`return` is not limited to one value. Separate them with commas, and unpack them into names on the way back out, in the same order.

````python
def check_record(value, limit):
    difference = value - limit
    percent = (value / limit) * 100
    return difference, percent

difference, percent = check_record(87, 100)
print(difference, percent)
````

## 5. The wall is the point, not a limitation

Because a function cannot reach outside itself, calling it can never accidentally
break something unrelated elsewhere in the program. Everything it needs comes in as a
parameter. Everything it produces goes out via `return`. That is the whole contract —
and it is what makes a function safe to call from anywhere.

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| using a variable created inside a function, outside it | `NameError` | `return` it, and catch the result outside |
| expecting a function to change a variable you passed in | the outside variable is unchanged | `return` the new value and reassign it yourself |
| assuming an error will warn you when a change is silently lost | no error at all — the bug is quiet | this is exactly why you `return` and reassign explicitly |
| unpacking into the wrong number of names | `ValueError: too many values to unpack` (or not enough) | the number of names must match the number of returned values, in order |
