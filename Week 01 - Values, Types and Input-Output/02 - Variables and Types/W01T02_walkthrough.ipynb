## 1. A variable is a name for a value

Read `=` as **"gets"**, not "equals".
`value = 23.7` means *put 23.7 into value*.

````python
record_id = "R-004"
value = 23.7
limit = 20

print(record_id)
print(value)
print(limit)
````

Once stored, use the **name** wherever you would use the value.

````python
print(record_id, "has value", value)
````

## 2. Types

Every value has a type. Ask Python with `type()`.

````python
print(type(record_id))
print(type(value))
print(type(limit))
````

| Type | Means |
|---|---|
| `str` | text |
| `int` | whole number |
| `float` | number with a decimal point |

### The thing that catches everyone

These two print identically. They are **not** the same.

````python
text_twenty = "20"
number_twenty = 20

print(text_twenty)
print(number_twenty)
````

````python
print(type(text_twenty))
print(type(number_twenty))
````

## 3. Naming

Names should say what the value *is*.

Rules: lowercase, underscores between words, no spaces, cannot start with a digit, cannot be a Python word such as `print`, `if` or `class`.

````python
# hard to read later
v = 23.7
l = 20

# easy to read later
sensor_value = 23.7
sensor_limit = 20

print(v - l)
print(sensor_value - sensor_limit)
````

## 4. Reassigning

A variable can be given a new value. The old one is gone.

````python
checks_done = 0
print(checks_done)

checks_done = 1
print(checks_done)

checks_done = 2
print(checks_done)
````

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| `23.7 = value` | `SyntaxError` | the name goes on the **left** |
| `my value = 3` | `SyntaxError` | no spaces — `my_value` |
| using a name you never set | `NameError` | assign it before you use it |
| `"20" + 5` | `TypeError` | text and number — convert first (Topic 4) |

> When something behaves oddly, **print the type**. It is almost always this.
