## 1. `def` creates it, calling runs it

Defining a function does not run it. Nothing happens until you call it — and you
can call it as many times as you like.

````python
def greet():
    print("Record check starting")

greet()
greet()
````

## 3. `return` sends a value back out

`print()` **shows** something. `return` **gives** something back to use.

````python
def add_tax(price):
    return price * 1.2

total = add_tax(100)
print(total)
````

      return (part / total) * 100

## 4. Why functions

One job each. Fix the calculation in one place, and every call is fixed — instead of
hunting down every copy you pasted it into.

A one-line string right after `def`, before any code, documents what the function
does. Python treats it specially — this is called a **docstring**.

````python
def percent_of(part, total):
    """Return part as a percentage of total."""
    return (part / total) * 100

help(percent_of)
````

Write a function `report_line(label, value)` that **returns** a string shaped like

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| forgetting `()` when calling — `greet` not `greet()` | nothing happens, or you see `<function greet at ...>` | add the brackets |
| `print`ing inside a function instead of `return`ing | works once, breaks the moment you need the value again | decide: does this function show, or give back? |
| expecting a value without catching it — `add_tax(100)` on its own | the returned value is thrown away | `total = add_tax(100)` |
| defining a function but never calling it | nothing happens, no error | `def` alone does nothing — you must call it |
| a docstring that just repeats the function's name | technically legal, tells you nothing | say what it *does*, not what it's called |
